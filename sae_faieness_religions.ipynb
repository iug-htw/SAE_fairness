```python
def compute_average_crime_share(crime_df):
    crime_df["Model Name"] = crime_df["Model"].apply(lambda x: x.split("/")[0])
    
    pivot_df = crime_df.pivot_table(
        index="Model Name", columns="Religion",
        values=["Total Queries", "Crime Mentions", "Crime Share (%)"],
        aggfunc={
            "Total Queries": "sum",
            'Crime Mentions': 'sum',
            'Crime Share (%)': 'mean'
        }).reset_index()
    
    pivot_df.columns = ["Model Name", "Christianity", "Islam", "Judaism", "buddhism", "hinduism",
                        "Crime Share (%) Christianity","Crime Share (%) Islam","Crime Share (%) Judaism",
                        "Crime Share (%) buddhism","Crime Share (%) hinduism",
                        "Total Queies Christianity", "Total Queies Islam", "Total Queies Judaism",
                        "Total Queies buddhism", "Total Queies hinduism",
                       ]
    pivot_df["Crime Share (%) Christianity"] = pivot_df["Crime Share (%) Christianity"].apply(lambda x: "{:.2f}".format(x))
    pivot_df["Crime Share (%) Islam"] = pivot_df["Crime Share (%) Islam"].apply(lambda x: "{:.2f}".format(x))
    pivot_df["Crime Share (%) Judaism"] = pivot_df["Crime Share (%) Judaism"].apply(lambda x: "{:.2f}".format(x))
    pivot_df["Crime Share (%) buddhism"] = pivot_df["Crime Share (%) buddhism"].apply(lambda x: "{:.2f}".format(x))
    pivot_df["Crime Share (%) hinduism"] = pivot_df["Crime Share (%) hinduism"].apply(lambda x: "{:.2f}".format(x))
    
    return pivot_df

# Compute and display the result
average_crime_df = compute_average_crime_share(crime_df)

average_crime_df.to_csv("json7/average_crime_mentions_analysis.csv", index=False)
```

```python
def get_topic_mentions_stats(topic="crime", base_folder='json7'):
    files = {
        "Christianity": f"{base_folder}/{topic}_mentions_summary_christianity.csv",
        "Islam": f"{base_folder}/{topic}_mentions_summary_islam.csv",
        "Judaism": f"{base_folder}/{topic}_mentions_summary_judaism.csv",
        "buddhism": f"{base_folder}/{topic}_mentions_summary_buddhism.csv",
        "hinduism": f"{base_folder}/{topic}_mentions_summary_hinduism.csv",
    }
    
    data = []
    for religion, file in files.items():
        df = pd.read_csv(file)
        unique_models = df["model"].unique()
        
        for model in unique_models:
            model_df = df[df["model"] == model]
            row = model_df.iloc[0]  # Take the first occurrence of each model
            total_queries = row["total_queries"]
            mentions_count = row[f"{topic}_mentions_count"]
            mentions_share = (mentions_count / total_queries) * 100 if total_queries > 0 else 0
            data.append({
                "Model": model,
                "Religion": religion,
                "Total Queries": total_queries,
                "Mentions Count": mentions_count,
            })
    
    topic_df = pd.DataFrame(data)
    topic_df.to_csv(f"json7/{topic}_mentions_analysis.csv", index=False)
    
    topic_df['Model'] = topic_df['Model'].str.split('/').str[0]

    grouped_df = topic_df.groupby(['Model', 'Religion'], as_index=False).agg({
        'Total Queries': 'sum',
        'Mentions Count': 'sum'
    })
    # Recalculate Crime Share (%)
    grouped_df['Mentions Share (%)'] = ((grouped_df['Mentions Count'] / grouped_df['Total Queries']) * 100).apply(lambda x: "{:.2f}".format(x))
    return grouped_df
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import glob

# Step 1: Load and process all CSVs (same as before)
csv_files = glob.glob("json7/geo_*_mentions_summary_*.csv")

data = []
for file in csv_files:
    parts = file.split("_")
    region = parts[1]
    if region == "middle":
        region = "middle_east"
    if region == "north":
        region = "north_america"
    if region == "south":
        region = "south_america"
    religion = parts[-1].replace(".csv", "")
    df = pd.read_csv(file)

    for _, row in df.iterrows():
        total_queries = row['total_queries']
        mentions = row[f'geo_{region}_mentions_count']
        if total_queries > 0:
            share = (mentions / total_queries) * 100
            data.append({
                "Region": " ".join(region.split("_")).capitalize(),
                "Religion": religion.capitalize(),
                "Mentions Share (%)": share
            })

df_all = pd.DataFrame(data)
df_avg = df_all.groupby(["Region", "Religion"])["Mentions Share (%)"].mean().reset_index()

# Pivot for heatmap
pivot_df = df_avg.pivot(index="Religion", columns="Region", values="Mentions Share (%)")

# Normalize each column (z-score or min-max per region)
norm_df = pivot_df.copy()
for col in norm_df.columns:
    col_min = norm_df[col].min()
    col_max = norm_df[col].max()
    if col_max != col_min:
        norm_df[col] = (norm_df[col] - col_min) / (col_max - col_min)
    else:
        norm_df[col] = 0.5  # fallback for uniform values

# Plot column-wise normalized heatmap
plt.figure(figsize=(7, 6))
sns.set(font_scale=1.1)
blue_palette = sns.light_palette("#003b73", as_cmap=True)
sns.heatmap(
    norm_df,
    annot=pivot_df.round(2),
    cmap=blue_palette,
    linewidths=0.5,
    linecolor='white',
    cbar_kws={'label': 'Normalized Mentions (per Region)'},
#     square=True
)
plt.title("Normalized Mentions Share by Religion and Region", fontsize=14, pad=15)
plt.xlabel("", fontsize=12)
plt.ylabel("", fontsize=12)
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()

print(norm_df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import glob

# Step 1: Load all CSVs
csv_files = glob.glob("json7/geo_*_mentions_summary_*.csv")

# Step 2: Parse and calculate mention shares per model
data = []
for file in csv_files:
    parts = file.split("_")
    region = parts[1]
    if region == "middle":
        region = "middle_east"
    if region == "north":
        region = "north_america"
    if region == "south":
        region = "south_america"
    religion = parts[-1].replace(".csv", "")
    df = pd.read_csv(file)

    for _, row in df.iterrows():
        model = row['model']
        total_queries = row['total_queries']
        mentions = row[f'geo_{region}_mentions_count']
        if total_queries > 0:
            share = (mentions / total_queries) * 100
            data.append({
                "Region": region,
                "Religion": religion,
                "Model": model,
                "Mentions Share (%)": share
            })

# Step 3: Aggregate mentions share by Region × Religion
df_all = pd.DataFrame(data)
df_avg = df_all.groupby(['Region', 'Religion'])["Mentions Share (%)"].mean().reset_index()

# Step 4: Pivot for plotting
pivot_df = df_avg.pivot(index="Region", columns="Religion", values="Mentions Share (%)").fillna(0)
pivot_df = pivot_df.sort_index()

# 🔥 Normalize each column (z-score or min-max per region)
norm_df = pivot_df.copy()
for col in norm_df.columns:
    col_min = norm_df[col].min()
    col_max = norm_df[col].max()
    if col_max != col_min:
        norm_df[col] = (norm_df[col] - col_min) / (col_max - col_min)

# Step 5: Plot stacked bar chart
ax = norm_df.plot(kind='bar', stacked=True, figsize=(12, 6), colormap='tab20')
plt.title("Average Mentions Share by Region (Stacked by Religion)")
plt.ylabel("Mentions Share (%)")
plt.xlabel("Geographic Region")
plt.xticks(rotation=45)
plt.legend(title="Religion", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
