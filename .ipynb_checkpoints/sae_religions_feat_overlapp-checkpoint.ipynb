# Religion–Violence–Geography with SAEs
## III. Latent Feature Overlap Analysis
---

This notebook contains the **latent feature overlap analysis** 
for the experiments described in:

**"Mechanistic Interpretability with SAEs: Probing Religion, Violence, and Geography in Large Language Models"**  
*(AEQUITAS Workshop @ ECAI 2025)*

It provides a reproducible workflow for:
- Intra-group overlaps: Do prompts from the same religion activate a shared set of features? (RQ1)  
- Inter-group overlaps: To what extent do religion-related prompts overlap with violence-related prompts? (RQ2)  
- Summary statistics and visualization of overlaps across models (contributing to RQ4)

These steps provide the overlap-based results used to address **RQ1 (intra-group coherence)**, 
**RQ2 (religion–violence associations)**, and **RQ4 (variation across models)**.  

**Note:**  
- For the **data collection and preliminary feature analysis**, refer to the **Data Collection & Preliminary Analysis notebook**.  
- For the **semantic probing of activation texts**, refer to the **Semantic Analysis notebook**.

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("json7/latent_feature_cosine_similarities.csv")

# Set group names as index
df.set_index("Group Name", inplace=True)

# Convert to float (in case some entries are "NA")
df = df.apply(pd.to_numeric, errors='coerce')

# Normalize each column (z-score or min-max per region)
norm_df = df.copy()
for col in norm_df.columns:
    col_min = norm_df[col].min()
    col_max = norm_df[col].max()
    if col_max != col_min:
        norm_df[col] = (norm_df[col] - col_min) / (col_max - col_min)
    else:
        norm_df[col] = 0.5  # fallback for uniform values

# Plot column-wise normalized heatmap
plt.figure(figsize=(10, 8))
sns.set(font_scale=1.1)
blue_palette = sns.light_palette("#003b73", as_cmap=True)
sns.heatmap(
    norm_df,
    annot=df.round(2),
    cmap=blue_palette,
    linewidths=0.5,
    linecolor='white',
    cbar_kws={'label': 'Cosine Similarity'},
#     square=True
)
plt.title("Inter-group Cosine Similarities")
plt.xlabel("Model:SAE")
plt.ylabel("Group Pair")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
