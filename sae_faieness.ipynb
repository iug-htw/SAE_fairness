```python
files = {
    "Christianity": "json5/crime_mentions_summary_christianity.csv",
    "Islam": "json5/crime_mentions_summary_islam.csv",
    "Judaism": "json5/crime_mentions_summary_judaism.csv"
}

crime_data = []
for religion, file in files.items():
    df = pd.read_csv(file)
    unique_models = df["model"].unique()
    
    for model in unique_models:
        model_df = df[df["model"] == model]
        row = model_df.iloc[0]  # Take the first occurrence of each model
        total_queries = row["total_queries"]
        crime_mentions = row["crime_mentions_count"]
        crime_share = (crime_mentions / total_queries) * 100 if total_queries > 0 else 0
        crime_data.append({
            "Model": model,
            "Religion": religion,
            "Total Queries": total_queries,
            "Crime Mentions": crime_mentions,
            "Crime Share (%)": round(crime_share, 2)
        })

crime_df = pd.DataFrame(crime_data)
crime_df.to_csv("crime_mentions_analysis.csv", index=False)
```

```python
# Step 2: Compute the average crime share per religion
def compute_average_crime_share(crime_df):
    crime_df["Model Name"] = crime_df["Model"].apply(lambda x: x.split("/")[0])
    
    pivot_df = crime_df.pivot_table(
        index="Model Name", columns="Religion",
        values=["Total Queries", "Crime Mentions", "Crime Share (%)"],
        aggfunc={
            "Total Queries": "sum",
            'Crime Mentions': 'sum',
            'Crime Share (%)': 'mean'
        }).reset_index()
    
    pivot_df.columns = ["Model Name", "Christianity", "Islam", "Judaism",
                        "Crime Share (%) Christianity","Crime Share (%) Islam","Crime Share (%) Judaism",
                        "Total Queies Christianity", "Total Queies Islam", "Total Queies Judaism",
                       ]
    pivot_df["Crime Share (%) Christianity"] = pivot_df["Crime Share (%) Christianity"].apply(lambda x: "{:.2f}".format(x))
    pivot_df["Crime Share (%) Islam"] = pivot_df["Crime Share (%) Islam"].apply(lambda x: "{:.2f}".format(x))
    pivot_df["Crime Share (%) Judaism"] = pivot_df["Crime Share (%) Judaism"].apply(lambda x: "{:.2f}".format(x))
    
    return pivot_df

# Compute and display the result
average_crime_df = compute_average_crime_share(crime_df)

average_crime_df.to_csv("average_crime_mentions_analysis.csv", index=False)
```
